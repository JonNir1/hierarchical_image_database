Compares the two cohorts' behavioral RDMs directly and through weighted-SMACOF MDS embeddings, for **retained subjects only** (production cohort, `status == "full data"`, `task_version >= 4.0`).<br>

distance vector with NaN for the ~35-37% of pairs no retained subject in that cohort ever judged.

## 1. Structural comparison via MDS

### 1.1 Weighted-SMACOF embeddings (R `smacof` backend)

Metric SMACOF via R's `smacof::mds()` (`SpAM_Simulations/core/multi_dimensional_scaling.py`'s
`run_mds()`), with binary observed/unobserved weights -- missing pairs are simply excluded from the
fit, with no imputation. One fit per dimensionality 2..`MAX_DIMS` per cohort
(28 fits total); each real-scale fit takes ~15-30s, so results are cached to
`analysis/results/pre_vs_post_shine/mds_cache/` (one `.npz` per cohort x dimension) and re-runs of this
notebook reuse the cache instead of refitting.

````python
MAX_DIMS = 15
MDS_DIMS = list(range(2, MAX_DIMS + 1))

mds_results = {
    c: run_mds_sweep(rdm[c], MDS_DIMS, c, MDS_CACHE_DIR, max_iters=1000, convergence_tol=1e-6)
    for c in COHORTS
}
for c in COHORTS:
    print(f"  {COHORT_LABEL[c]}: {len(mds_results[c])} dimensionalities ready (cached under {MDS_CACHE_DIR})")
````

### 1.2 Stress scree plot

Weighted-SMACOF stress-1 across dimensionality, one trace per cohort -- the validated
(no-imputation) dimensionality diagnostic.

````python
fig_stress_scree(mds_results, MDS_DIMS).show()
````

### 1.3 Eigenvalue spectrum of the MDS embeddings

Each SMACOF embedding is a configuration of 725 points in `d` dimensions. Its eigenvalue spectrum is the
variance along each of the `d` principal axes of that configuration (equivalently, the eigenvalues of the
double-centred Gram matrix of the embedding's own pairwise distances), here as a proportion of the total
variance so that the pre and post fits, whose overall scales are arbitrary, are comparable. Nothing is
imputed: the fit ignores unobserved pairs, so the spectrum describes the fitted geometry itself. Three
dimensionalities are shown (3, 9, 15), sharing the y-axis. A spectrum that falls steeply means a few axes
carry most of the variance; a gap between the cohorts' curves means they spread variance across the axes
differently.

````python
SPECTRUM_DIMS = (3, 9, 15)
spectra = {d: {c: embedding_eigenspectrum(mds_results[c][d]["conf"]) for c in COHORTS} for d in SPECTRUM_DIMS}
fig_embedding_spectrum(spectra).show()
````

### 1.4 Pre- vs. post-SHINE structural agreement

For the raw RDMs and for each embedding dimensionality, do the pre- and post-SHINE cohorts agree
on the *same* similarity structure? Kendall tau and Spearman rho are computed on the raw RDMs too
(restricted to jointly-covered pairs), but Procrustes m^2 needs a complete point configuration, so
it's only defined once an MDS embedding exists (`NaN` for "raw RDM" below). Procrustes m^2 is
shown as `1 - m^2` so all three traces read "bigger = better".

````python
agreement: dict[str, dict] = {}
agreement["raw RDM"] = {**masked_rank_correlation(rdm["pre"], rdm["post"]), "procrustes_m2": None}
for d in MDS_DIMS:
    ks = masked_rank_correlation(mds_results["pre"][d]["confdist"], mds_results["post"][d]["confdist"])
    m2 = procrustes_m2(mds_results["pre"][d]["conf"], mds_results["post"][d]["conf"])
    agreement[str(d)] = {**ks, "procrustes_m2": m2}

agreement_table = pd.DataFrame(agreement).T[["kendall", "spearman", "procrustes_m2", "n_overlap"]]
agreement_table
````

````python
fig_cross_cohort_agreement(agreement, MDS_DIMS).show()
````

### 1.5 Within- vs. between-cohort agreement at a matched group size

The raw-RDM agreement in 1.4 needs a reference: how well do two groups of subjects agree when the cohort is
not in question? Each draw takes two disjoint groups of 38 subjects (half of the smaller cohort) from the pre
cohort and two from the post cohort, and correlates them (Kendall tau and Spearman rho on jointly covered
pairs): pre with pre and post with post (the within-cohort noise ceiling at this group size, the logic of the
split-half bound in `aggregation_comparison.ipynb`) and pre with post (between cohorts; the mean of two
pairings, group 1 with group 1 and group 2 with group 2). 100 random splits per condition.

The values are lower than the pre-post agreement of 1.4 because the groups here are half as large (38 instead
of 76 or 79 subjects). They should not be extrapolated to the full sample with the Spearman-Brown formula:
on these sparse RDMs extra subjects mostly add covered pairs rather than repeated ratings per pair, so the
correlation grows slowly with group size. The draws reuse the same subjects, so the intervals describe
variation across splits, not sampling uncertainty; the inferential tests are in sections 2 and 3.

````python
N_JOBS = 6

cohorts = load_retained_cohorts(cache_dir)
subject_ids = {c: cohorts["participant_ids"][c].tolist() for c in COHORTS}
trials_by_subject, n_conds = parse_subject_trials(cohorts["df_trials"], subject_ids)
rdm_builder = partial(rdm_from_subject_trials, trials_by_subject=trials_by_subject, n_conds=n_conds)

within_between = within_between_agreement(
    rdm_builder, subject_ids["pre"], subject_ids["post"], n_draws=100, seed=0, n_jobs=N_JOBS,
    cache_path=PERMUTATION_CACHE_DIR / "within_between_seed0.npz",
)
condition_labels = {"within_a": "Pre vs. pre", "within_b": "Post vs. post", "between": "Pre vs. post"}
pd.DataFrame({
    statistic: {
        condition_labels[condition]: "{mean:.3f} [{low:.3f}, {high:.3f}]".format(**within_between[condition][statistic])
        for condition in within_between["conditions"]
    }
    for statistic in ("kendall", "spearman")
})
````

````python
fig_within_between(within_between, "kendall", statistic_label="Kendall tau").show()
fig_within_between(within_between, "spearman", statistic_label="Spearman rho").show()
````

## 2. Cohort-label permutation test of pre-post agreement

Section 1.4 reports how well the two cohorts' RDMs agree (raw RDM tau 0.15, rho 0.22). With over 100,000
jointly rated pairs any positive correlation is trivially significant, so the informative question is
whether the two cohorts agree *less than exchangeable cohorts would*, i.e. whether SHINE changes the
similarity structure beyond sampling noise. The test, per statistic (Kendall tau, Spearman rho on
jointly covered pairs):

1. Observed: build the pre RDM and post RDM ("simple" aggregation) and correlate them.
2. Pool all retained subjects and shuffle the cohort label at the subject level (group sizes kept).
3. Rebuild both RDMs from the shuffled groups and correlate them. Repeat `N_PERM` times.
4. One-sided p = (1 + number of shuffled values <= observed) / (1 + `N_PERM`).

The null already carries the same noise level, sample size and pair coverage, so its mean is the
agreement expected if the cohorts were identical. This tests for a difference, not for equivalence: a
non-small p means no detectable difference, not proof of none. It assumes subjects are exchangeable
across cohorts apart from SHINE. The null draws are cached under
`analysis/results/pre_vs_post_shine/permutation_cache/`.

````python
N_PERM = 250

permutation = permutation_test_cohort_agreement(
    rdm_builder, subject_ids["pre"], subject_ids["post"], n_perm=N_PERM, seed=0, n_jobs=N_JOBS,
    cache_path=PERMUTATION_CACHE_DIR / f"cohort_agreement_n{N_PERM}_seed0.npz",
)
pd.DataFrame({key: permutation[key] for key in ("kendall", "spearman")}).T.round(4)
````

````python
fig_permutation_null(permutation, "kendall", statistic_label="Kendall tau").show()
fig_permutation_null(permutation, "spearman", statistic_label="Spearman rho").show()
````

## 3. Equivalence of pre-post agreement

Section 2 tests whether the cohorts agree *less* than exchangeable cohorts would; a non-small p there is
not evidence that they agree as much. This section tests equivalence directly. The effect is

    D = (observed agreement) - (mean of the label-permutation null)

for Kendall tau and Spearman rho, where "agreement" is the rank correlation between the pre and post RDMs on
jointly covered pairs, and the null mean is the agreement expected if the cohorts were identical (section 2).
D = 0 means the cohorts agree exactly as much as two exchangeable groups would. The question is whether D is
within a small bound of 0.

**Where the 90% CI of D comes from: a delete-a-group jackknife.** The CI needs the sampling uncertainty of D,
i.e. how much D would change with a different sample of subjects. The spread of the permutation null cannot
provide it: every shuffle reuses the same subjects, so it understates this uncertainty. The jackknife measures
it by recomputing D on many slightly smaller samples, each leaving out a different slice of subjects, and
looking at how much D moves. "Delete-one" would drop one subject at a time (155 recomputations, each with its
own null); "delete-a-group" drops a whole group at a time, so only G = 20 recomputations are needed.

```
D_obs = agreement(pre, post) - mean(null)               # null: the 250 label shuffles of section 2

split the pre subjects into G = 20 random groups          # about 4 subjects per group
split the post subjects into G = 20 random groups

for g in 1..G:
    pre_g  = all pre subjects except pre group g          # about 95% of the subjects remain
    post_g = all post subjects except post group g
    null_g = mean of agreement(A, B) over 10 shuffles of the pooled subjects of (pre_g, post_g),
             each shuffle re-splitting them into groups of the same sizes as pre_g and post_g
    D_g    = agreement(pre_g, post_g) - null_g

SE     = sqrt( (G - 1) / G  *  sum over g of (D_g - mean(D_1..D_G))^2 )
90% CI = D_obs  +/-  1.645 * SE
```

The factor (G - 1) / G corrects for the replicates being highly overlapping (each keeps about 95% of the
subjects), which makes them vary less than independent samples would. Only 10 shuffles per leave-out are used
for `null_g`; this adds a little noise to each D_g, which the SE absorbs.

**The bound.** As for the other equivalence tests, it is the smallest bound that would show equivalence with
80% power if D were truly 0: (1.645 + 1.28) x SE (z(0.95) + z(0.90)), rounded up to 0.01 in tau/rho units. D is judged equivalent
when the whole 90% CI lies inside the bound.

````python
equivalence_agreement = equivalence_test_cohort_agreement(
    rdm_builder, subject_ids["pre"], subject_ids["post"], permutation,
    n_groups=20, n_null=10, seed=0, n_jobs=N_JOBS,
    cache_path=PERMUTATION_CACHE_DIR / "cohort_agreement_jackknife_g20_n10_seed0.npz",
)
for statistic, result in equivalence_agreement.items():
    print(f"{statistic}: jackknife SE={result['se']:.4f}, 80%-power sensitivity={result['sensitivity']:.4f}")
    print_equivalence_line(result)
pd.DataFrame(equivalence_agreement).T[["effect", "ci_low", "ci_high", "se", "sensitivity", "bound", "equivalent"]]
````

````python
fig_equivalence_forest(
    {"Kendall tau": equivalence_agreement["kendall"], "Spearman rho": equivalence_agreement["spearman"]},
    x_title="Observed minus exchangeable expectation, 90% CI", title="Equivalence: pre-post RDM agreement",
).show()
````
