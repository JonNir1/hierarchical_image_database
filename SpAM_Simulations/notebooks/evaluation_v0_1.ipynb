# SpAM Simulation & MDS Evaluation Pipeline

This notebook provides an end-to-end evaluation pipeline for simulated Spatial Arrangement Method (SpAM) experiments. It is designed to test whether specific experimental configurations (number of subjects, trials, and noise levels) can successfully reconstruct a high-dimensional ground-truth semantic space using Multidimensional Scaling (MDS).

## Pipeline Overview

**1. Simulation Initialization & Loading**
* Loads or generates synthetic subject data based on a defined Ground Truth (GT) distance matrix.
* Sweeps across multiple experimental parameters: Subject count ($N$), trials per subject, images per trial, and human noise variability (scaled to the GT standard deviation).

**2. Graph Coverage & Connectivity Analysis**
* Evaluates the theoretical viability of the experimental configurations before running MDS.
* Calculates **Image Coverage** and **Pair Coverage** percentages.
* Calculates **P[Connected]**: The probability that the sampled image-pairs form a single, fully connected graph (a strict mathematical prerequisite for MDS).

**3. Pre-MDS Data Stability (Reliability)**
* Measures the stability of the raw simulated measurements.
* Calculates the Spearman rank correlation between distance matrices generated by different iterations of the *same* experimental configuration to establish a baseline data reliability ceiling.

**4. Multidimensional Scaling (MDS) Execution**
* Streams the simulated distance matrices into R's `smacof` algorithm via a memory-safe generator.
* Uses weighted SMACOF to properly handle sparse/missing data (unobserved pairs).
* Fits the data across a range of target dimensions ($D$) to simulate real-world exploratory analysis, saving iterations securely to disk (`mds_results.pkl`).

**5. MDS Diagnostics & Final Evaluation**
* **Failure Tracking:** Identifies convergence failures or disconnected components.
* **Scree Plots:** Visualizes MDS Stress versus Dimensionality to help identify the "elbow" and optimal target dimensionality.
* **Embedding Stability:** Calculates the final Spearman correlation of the reconstructed MDS distances (`confdist`) across simulation iterations to determine which experimental configurations yield mathematically stable and reproducible semantic maps.

## Evaluate Simulation
### Coverage
We check three coverage scores for the simulated data:
1. **Image Coverage**: the percentage of images that were observed at least once across all subjects and trials.
2. **Pair Coverage**: the percentage of image-pairs that were observed at least once across all subjects and trials.
3. **P[_is connected_]**: the probability that the graph of observed image-pairs is fully connected (i.e. has only 1 connected component). We check this because the `MDS` algorithm requires a fully connected graph to run, so if we have a low probability of getting a fully connected graph for a given experimental configuration, that configuration may not be suitable for running MDS on.

```python
coverage = pipeline.compute_coverage_table(sim).rename(columns={"num_subjects": "n_subjects"})
coverage["is_connected"] = coverage["num_connected_components"] == 1.0
coverage = coverage.set_index(
    ["n_subjects", "trials_per_subject", "images_per_trial",
     "subjects_noise_scale", "subjects_noise_df", "rep"]
)

# coverage scores are not affected by subject noise, so we average across those parameters
coverage_summary = (
    coverage.groupby(["n_subjects", "trials_per_subject", "images_per_trial"])
    .agg(
        num_reps=("img_coverage", "size"),
        percent_img_coverage_mean=("img_coverage", "mean"),
        percent_img_coverage_sem=("img_coverage", "sem"),
        percent_pair_coverage_mean=("pair_coverage", "mean"),
        percent_pair_coverage_sem=("pair_coverage", "sem"),
        p_is_connected_mean=("is_connected", "mean"),
        p_is_connected_sem=("is_connected", "sem"),
    )
    .sort_index()
    .reset_index()
)
```

```python
ROW_TITLES = {
    "% IMG COVERAGE": "percent_img_coverage",
    "% PAIR COVERAGE": "percent_pair_coverage",
    "P[CONNECTED]": "p_is_connected",
}
COL_TITLES = {k: f"{k} Images per Trial" for k in coverage_summary["images_per_trial"].unique()}
coverage_fig = make_subplots(
    rows=len(ROW_TITLES), cols=len(COL_TITLES),
    column_titles=list(COL_TITLES.values()),
    shared_xaxes=True, shared_yaxes=True,
    x_title="Number of Subjects",
    vertical_spacing=0.05, horizontal_spacing=0.025,
)
for c, (images_per_trial, col_title) in enumerate(COL_TITLES.items()):
    for r, (row_title, prefix) in enumerate(ROW_TITLES.items()):
        for i, tr in enumerate(coverage_summary["trials_per_subject"].unique()):
            color = px_colors.qualitative.Plotly[i]
            df = coverage_summary.loc[
                (coverage_summary["trials_per_subject"] == tr) & (coverage_summary["images_per_trial"] == images_per_trial),
            ]
            coverage_fig.add_trace(
                row=r + 1, col=c + 1, trace=go.Scatter(
                    x=df["n_subjects"],
                    y=df[f"{prefix}_mean"],
                    error_y=dict(type="data", array=df[f"{prefix}_sem"], visible=True),
                    name=str(tr), legendgroup=str(tr), showlegend=c == 0 and r == 0,
                    mode="lines+markers", line=dict(color=color),
                )
            )
        if c == 0:
            coverage_fig.update_yaxes(
                row=r + 1, col=c + 1,
                title=dict(text=row_title, font=dict(size=14, color='black'))
            )
del c, r, i, images_per_trial, col_title, row_title, prefix, tr, color, df

coverage_fig.update_layout(
    height=650, width=1500,
    title=dict(
        text="Coverage by Experimental Configuration", font=dict(size=20, color='black')
    ),
    legend=dict(
        title=dict(text="Trials per Subject"),
        orientation="h",
        x=1.0, xanchor="right", xref="paper",
        y=1.075, yanchor="bottom", yref="paper",
    ),
)
coverage_fig.show()
```

### Stability
We check the Spearman (rank) correlation between different iterations of the same experimental configuration to see how stable the results are.

```python
# Pre-MDS reliability: mean Spearman correlation between repetitions of each configuration.
PARAM_FIELDS = [
    "num_subjects", "trials_per_subject", "images_per_trial", "subjects_noise_scale", "subjects_noise_df"
]

correlations = (
    pipeline.compute_stability_table(sim)
    .dropna(subset=["spearman"])
    .groupby(PARAM_FIELDS)
    .agg(count=("spearman", "size"), r_mean=("spearman", "mean"), r_sem=("spearman", "sem"))
)
correlations.index = correlations.index.set_names(
    ["n_subjects", "trials_per_subject", "images_per_trial",
     "subjects_noise_scale", "subjects_noise_df"]
)
```

```python
ROW_TITLES = {tr: f"{tr} Trials per Subject" for tr in coverage_summary["trials_per_subject"].unique()}
COL_TITLES = {k: f"{k} Images per Trial" for k in coverage_summary["images_per_trial"].unique()}
corr_fig = make_subplots(
    rows=len(ROW_TITLES), cols=len(COL_TITLES),
    column_titles=list(COL_TITLES.values()),
    shared_xaxes=True, shared_yaxes=True,
    x_title="Number of Subjects",
    vertical_spacing=0.05, horizontal_spacing=0.025,
)
for c, (images_per_trial, col_title) in enumerate(COL_TITLES.items()):
    for r, (trials_per_subject, row_title) in enumerate(ROW_TITLES.items()):
        subset = correlations.loc[
            (correlations.index.get_level_values("images_per_trial") == images_per_trial) &
            (correlations.index.get_level_values("trials_per_subject") == trials_per_subject)
        ]
        noise_scales = subset.index.get_level_values("subjects_noise_scale").unique()
        noise_dfs = subset.index.get_level_values("subjects_noise_df").unique()
        for i, (noise_scale, noise_df) in enumerate(product(noise_scales, noise_dfs)):
            name = f"Scale={noise_scale}<br>DFs={noise_df}"
            color = px_colors.qualitative.Plotly[i % len(px_colors.qualitative.Plotly)]
            df = subset.loc[
                (subset.index.get_level_values("subjects_noise_scale") == noise_scale) &
                (subset.index.get_level_values("subjects_noise_df") == noise_df)
            ]
            corr_fig.add_trace(
                row=r + 1, col=c + 1, trace=go.Scatter(
                    x=df.index.get_level_values("n_subjects"),
                    y=df["r_mean"],
                    error_y=dict(type="data", array=df["r_sem"], visible=True),
                    name=name, legendgroup=name, showlegend=c == 0 and r == 0,
                    mode="lines+markers", line=dict(color=color),
                )
            )
        if c == 0:
            corr_fig.update_yaxes(
                row=r + 1, col=c + 1,
                title=dict(text=row_title, font=dict(size=10, color='black'))
            )
del c, r, i, images_per_trial, col_title, trials_per_subject, row_title, noise_scale, noise_df, name, color, df

corr_fig.update_layout(
    height=650, width=1500,
    title=dict(
        text="Stability (Spearman Correlation) by Experimental Configuration",
        font=dict(size=20, color='black')
    ),
    legend=dict(
        title=dict(text="Subject Noise Parameters"),
    ),
)
corr_fig.show()
```

## Evaluate Embeddings
### Stability
We want to check how "stable" our embeddings are across different iterations of the same experimental configuration. To do this, we calculate the Spearman (rank) correlation between the pairwise distances of the embeddings from different iterations of the same experimental configuration.<br>
<br>
_Why use the Spearman correlation?_<br>
- _Cosine Similarity_: We don't care about the angle between embeddings, just that we maintain their distances relative to each other. We should allow for scaling and rotation of the embeddings, which would change the cosine similarity but not the relative distances between points.<br>
- _Pearson Correlation_: We don't care about the exact distances between points, just that they maintain their relative distances to each other. We should allow for scaling and rotation of the embeddings, which would change the Pearson correlation but not the Spearman correlation.<br>
- _Procrustes Analysis_: This is a more complex method that involves finding the optimal scaling and rotation to align two embeddings before calculating their similarity. This could be a good option, but it is more computationally intensive and may not be necessary for our purposes. The Spearman correlation is a simpler and more interpretable metric that still captures the stability of the embeddings across iterations.<br>

```python
# Post-MDS embedding stability: mean Spearman agreement of reconstructed distances across reps.
embedding_corrs_df = (
    pipeline.compute_embedding_stability(store)
    .rename(columns={"n_reps": "N", "mean_spearman": "r_mean", "sem_spearman": "r_sem"})
)
```

```python
ROW_TITLES = {tr: f"{tr} Trials per Subject" for tr in embedding_corrs_df["trials_per_subject"].unique()}
COL_TITLES = {k: f"{k} Images per Trial" for k in embedding_corrs_df["images_per_trial"].unique()}
embedded_corr_fig = make_subplots(
    rows=len(ROW_TITLES), cols=len(COL_TITLES),
    column_titles=list(COL_TITLES.values()),
    shared_xaxes=True, shared_yaxes=True,
    x_title="DIMENSIONS",
    vertical_spacing=0.05, horizontal_spacing=0.025,
)
for c, (images_per_trial, col_title) in enumerate(COL_TITLES.items()):
    for r, (trials_per_subject, row_title) in enumerate(ROW_TITLES.items()):
        subset = embedding_corrs_df.loc[
            (embedding_corrs_df["images_per_trial"] == images_per_trial) &
            (embedding_corrs_df["trials_per_subject"] == trials_per_subject)
        ]
        noise_scales = sorted(embedding_corrs_df["subjects_noise_scale"].unique())
        noise_dfs = sorted(embedding_corrs_df["subjects_noise_df"].unique())
        n_subjs = sorted(embedding_corrs_df["num_subjects"].unique())
        for i, (noise_scale, noise_df, n_subj) in enumerate(product(noise_scales, noise_dfs, n_subjs)):
            name = f"Scale={noise_scale}<br>DFs={noise_df}<br>N={n_subj}"
            color = px_colors.qualitative.Plotly[i % len(px_colors.qualitative.Plotly)]
            df = subset.loc[
                    (embedding_corrs_df["subjects_noise_scale"] == noise_scale) &
                    (embedding_corrs_df["subjects_noise_df"] == noise_df) &
                    (embedding_corrs_df["num_subjects"] == n_subj)
                ]
            embedded_corr_fig.add_trace(
                row=r + 1, col=c + 1, trace=go.Scatter(
                    x=df["ndim"],
                    y=df["r_mean"],
                    error_y=dict(type="data", array=df["r_sem"].fillna(0), visible=True),
                    name=name, legendgroup=name, showlegend=c == len(COL_TITLES)-1 and r == len(ROW_TITLES)-1,
                    mode="lines+markers", line=dict(color=color),
                    hovertemplate=f"{name}<br>" + "NDIM=%{x}<br>Spearman's R=%{y:.4f}",
                )
            )
        if c == 0:
            embedded_corr_fig.update_yaxes(
                row=r + 1, col=c + 1,
                title=dict(text=row_title, font=dict(size=10, color='black'))
            )
del c, r, i, images_per_trial, col_title, trials_per_subject, row_title, noise_scale, noise_df, name, color,

embedded_corr_fig.update_layout(
    height=650, width=1500,
    title=dict(
        text="MDS-Stability (Spearman Correlation) by Experimental Configuration",
        font=dict(size=20, color='black')
    ),
    legend=dict(
        title=dict(text="Subject Noise Parameters"),
    ),
)
embedded_corr_fig.show()
```

#### Convergence

```python
def plot_convergence_by_dimension(convergence_df, ndim):
    """
    Filters convergence data by target dimensionality and generates a 100% stacked bar chart
    showing both aggregated and noise-split convergence statuses.
    """
    # 1. Filter for the target dimensionality
    df_filtered = convergence_df[convergence_df['ndim'] == ndim].copy()

    if df_filtered.empty:
        print(f"Warning: No data found for ndim = {ndim}")
        return None

    # 2. Dynamically determine layout based on unique noise scales
    noise_scales = sorted(df_filtered['subjects_noise_scale'].unique())
    num_cols = max(1, len(noise_scales))

    # Row 1 spans all columns, Row 2 creates an individual column for each noise scale
    specs = [[{"colspan": num_cols}] + [None] * (num_cols - 1),
             [{}] * num_cols]

    subplot_titles = ["Aggregated Mixture (All Noise Scales)"] + [f"Noise = {n}" for n in noise_scales]

    fig = make_subplots(
        rows=2, cols=num_cols,
        specs=specs,
        subplot_titles=subplot_titles,
        vertical_spacing=0.15
    )

    # 3. Configuration for traces
    statuses = ['converged', 'max_iters', 'disconnected']
    colors = {'converged': '#2ca02c', 'max_iters': '#ff7f0e', 'disconnected': '#d62728'}
    names = {'converged': 'Success', 'max_iters': 'Max Iterations', 'disconnected': 'Disconnected'}

    def add_stacked_bars(data, row, col, showlegend=False):
        # Aggregate counts by num_subjects
        grouped = data.groupby('num_subjects')[statuses].sum().reset_index()

        for status in statuses:
            if status in grouped.columns:
                fig.add_trace(
                    go.Bar(
                        x=grouped['num_subjects'],
                        y=grouped[status],
                        name=names[status],
                        marker_color=colors[status],
                        showlegend=showlegend, # Only show legend once to avoid duplicates
                        legendgroup=status,    # Links legend items across subplots so 1 click toggles all
                    ),
                    row=row, col=col
                )

    # 4. Add Main Plot (Aggregated)
    add_stacked_bars(df_filtered, row=1, col=1, showlegend=True)

    # 5. Add Subplots (Split by Noise Scale)
    for i, noise in enumerate(noise_scales):
        df_noise = df_filtered[df_filtered['subjects_noise_scale'] == noise]
        add_stacked_bars(df_noise, row=2, col=i+1, showlegend=False)

    # 6. Format layout (Enforce 100% Stacking via `barnorm`)
    fig.update_layout(
        barmode='stack',
        barnorm='percent',  # Converts raw counts to 100% percentages
        title=f"SpAM Convergence Status by Sample Size (Target D={ndim})",
        height=800,
        width=1200,
        legend=dict(orientation="h", yanchor="bottom", y=1.05, xanchor="right", x=1),
        template="plotly_white"
    )

    # Fix Axis Labels and Ranges
    fig.update_yaxes(title_text="Percentage (%)", range=[0, 100], row=1, col=1)
    fig.update_xaxes(title_text="Number of Subjects", type='category', row=1, col=1)

    for i in range(1, num_cols + 1):
        fig.update_yaxes(title_text="%", range=[0, 100], row=2, col=i)
        fig.update_xaxes(title_text="Number of Subjects", type='category', row=2, col=i)

    return fig
```

```python
# Convergence status counts for the focus configuration, built from the MDS store metadata.
_status_map = {"success": "converged", "max_iters": "max_iters", "disconnected": "disconnected"}
conv = mds_meta[
    (mds_meta["trials_per_subject"] == TRIALS_PER_SUBJECT)
    & (mds_meta["images_per_trial"] == IMAGES_PER_TRIAL)
    & (mds_meta["subjects_noise_scale"] != EXCLUDE_NOISE_SCALE)
    & (mds_meta["subjects_noise_df"] != EXCLUDE_NOISE_DF)
    & (mds_meta["ndim"].isin(EMBEDDING_DIMS))
].copy()
conv["status"] = conv["status"].map(_status_map).fillna("other_error")

convergence_statuses_df = (
    conv.groupby(["num_subjects", "subjects_noise_scale", "ndim"], observed=True)["status"]
    .value_counts()
    .unstack(fill_value=0)
    .reset_index(drop=False)
)
for _s in ["converged", "max_iters", "disconnected"]:
    if _s not in convergence_statuses_df.columns:
        convergence_statuses_df[_s] = 0

plot_convergence_by_dimension(convergence_statuses_df, ndim=EMBEDDING_DIMS[-1]).show()
plot_convergence_by_dimension(convergence_statuses_df, ndim=EMBEDDING_DIMS[0]).show()
```

#### Stability
##### Raw Measurement Stability
Spearman correlation between **generated** distance matrices - this gives us the noise ceiling of our data before we even run MDS.
##### Embedding Stability
Spearman correlation between **reconstructed** distance matrices (`confdist`) from MDS - this tells us how stable the final embeddings are across iterations experimental configuration.

```python
# Embedding stability (post-MDS) for the focus configuration, from the stored confdists.
focus = mds_meta[
    (mds_meta["trials_per_subject"] == TRIALS_PER_SUBJECT)
    & (mds_meta["images_per_trial"] == IMAGES_PER_TRIAL)
    & (mds_meta["subjects_noise_scale"] != EXCLUDE_NOISE_SCALE)
    & (mds_meta["subjects_noise_df"] != EXCLUDE_NOISE_DF)
    & (mds_meta["ndim"].isin(EMBEDDING_DIMS))
    & (mds_meta["status"].isin(["success", "max_iters"]))
    & (mds_meta["confdist_row"] >= 0)
].copy()

rows = []
for (n_subj, noise_scale, ndim), grp in focus.groupby(
    ["num_subjects", "subjects_noise_scale", "ndim"], observed=True
):
    confdists = [store.confdist(int(r)) for r in grp["confdist_row"]]
    rs = [spearmanr(a, b).statistic for a, b in combinations(confdists, 2)]
    rows.append({
        "num_subjects": n_subj, "noise_scale": noise_scale, "ndim": ndim,
        "count": len(rs),
        "r_mean": float(np.mean(rs)) if rs else np.nan,
        "r_sem": float(np.std(rs, ddof=1) / np.sqrt(len(rs))) if len(rs) > 1 else np.nan,
    })
embedding_corrs_summary_subset = pd.DataFrame(rows)
```
